```python
import pandas as pd
```

```python
file_path = "../1.DATASET/CMI_FINAL_OD.csv"
df=pd.read_csv(file_path) 
```

```python


# Seleziona colonne BIA + colonne aggiuntive
bia_cols = [col for col in df.columns if col.startswith("BIA")]
extra_cols = ["Physical-Weight", "Physical-Height", "Basic_Demos-Age", "Physical-BMI", "sii"]

# Verifica che le colonne extra esistano nel dataset
extra_cols_found = [col for col in extra_cols if col in df.columns]
missing = [col for col in extra_cols if col not in df.columns]
if missing:
    print(f"⚠️  Colonne non trovate nel dataset: {missing}")

cols_to_show = bia_cols + extra_cols_found

print(f"Colonne BIA trovate ({len(bia_cols)}): {bia_cols}")
print(f"Colonne extra trovate: {extra_cols_found}\n")

# Duplicati esatti su tutte le colonne BIA
n_dup = df.duplicated(subset=bia_cols).sum()
print(f"Righe duplicate (esatte su tutte le BIA): {n_dup}")
print(f"Percentuale: {n_dup / len(df) * 100:.2f}%\n")

if n_dup > 0:
    print("Righe duplicate — visualizzate con colonne BIA + fisiche:\n")
    righe_dup = (
        df[df.duplicated(subset=bia_cols, keep=False)]
        .sort_values(by=bia_cols)
        [cols_to_show]  # mostra solo le colonne di interesse
    )
    print(righe_dup.head(87).to_string(index=True))
```
